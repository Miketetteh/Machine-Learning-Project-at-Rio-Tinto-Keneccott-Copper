```python
# Using the melt function to stack multiple columns into one column
df_stacked = pd.melt(filtered_df, var_name='Column', value_name='node')
```

```python
df_stacked
```

```python
uq_id = df_stacked['Column'].unique()
uq_id
```

```python
df_stacked = df_stacked.to_numpy()
```

```python
df_stacked.shape
```

```python
el = el.to_numpy()
```

```python
layer = []
elev = []
x = 1
for i in range(len(uq_id)):
    for j in range(len(df_stacked)):
        if uq_id[i] == df_stacked[j][0]:
            layer.append(x)
            elev.append(el[i])
    x += 1
```

```python
layer = np.array(layer)
layer = layer.reshape(23120, 1)
elev = np.array(elev)
layer.shape
```

```python
df_stack = np.hstack((df_stacked, layer, elev))
```

```python
columns = ['column', 'node', 'layer',' elev']
df_stack = pd.DataFrame(df_stack, columns=columns)
```
