```python
# !!!!!!!!!!!!!!! CHANGE THESE EACH YEAR !!!!!!
path_hdrn_base = os.path.join(r'C:\Users\Michael.Tetteh\Desktop\hz_spfl')
path_hdrn =  r'C:\Users\Michael.Tetteh\Desktop\hz_spfl\intersect'       #os.path.join(path_hdrn_base, 'interset')
path_out = os.path.join(r'C:\Users\Michael.Tetteh\Desktop\output') 
fn_out   = 'ApexFS_00_AltA01k_HorzDrn_toGWV_Y' + str(year) + '.csv'
cols = ['node', 'layer', 'elev', 'dry']
```

```python
start_sp = (df_yrsp[df_yrsp['Yr'] == year]['SP'])     #int
end_sp = df_yrsp['SP'].max()
```

```python
list_node  = []
list_layer = []
list_elev  = []
list_dry   = []
```

```python
for subdir, dirs, files in os.walk(path_hdrn):
    for file in files: 
        ext = os.path.splitext(file)
        if ext[1] == '.dbf':
            print('reading ' + ext[0])
            dbf = Dbf5(path_hdrn + '\\' + file)
            df_hdrn = dbf.to_dataframe()     #[cols]
df_hdrn
```

```python
for subdir, dirs, files in os.walk(path_hdrn):
    for file in files: 
        ext = os.path.splitext(file)
        if ext[1] == '.dbf':
            print('reading ' + ext[0])
            dbf = Dbf5(path_hdrn + '\\' + file)
            df_hdrn = dbf.to_dataframe()[cols]

            df_hdrn['node'] = df_hdrn['node'].astype(int)
            df_hdrn['layer'] = df_hdrn['layer'].astype(int)
            df_hdrn['elev']  = df_hdrn['elev'] + refhd

            # remove duplicate nodes - shoule each have same attributes
            df_hdrn_unq = df_hdrn.drop_duplicates('node', inplace=True)

            listN = list(df_hdrn['node'])
            listL = list(df_hdrn['layer'])
            listE = list(df_hdrn['elev'])
            listD = list(df_hdrn['dry'])

            list_node.extend(listN)
            list_layer.extend(listL)
            list_elev.extend(listE)
            list_dry.extend(listD)

N = len(list_node)
df_hdrn_out = pd.DataFrame(data = {'node': list_node, 'layer': list_layer, 'elev': list_elev, 'dry': list_dry, 'reach': N * [reach], 'startSP': N * [start_sp], 'endSP': N * [end_sp], 'drncond': N*[drncond]})
df_hdrn_out.to_csv(path_out + '\\' + fn_out, index=False)
            
```
