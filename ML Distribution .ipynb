```python
import pandas as pd
import pickle

# UPDATE THIS FILE PATH TO the csv you want the code to predict, MAke sure the CSV has 4 columns, ID_PIEZO, TM_STAMP (hourly),WL_ELEV, and PIEZO_RL
new_data = pd.read_csv('fake_1.csv')


new_data['TM_STAMP'] = (pd.to_datetime(new_data['TM_STAMP']).view('int64') // 10**9)
new_data.dropna(subset=['PIEZO_RL'], inplace=True)

X_new = new_data[['ID_PIEZO', 'TM_STAMP', 'WL_Elev', 'PIEZO_RL']]


with open(r'K:\Mine\TechnicalServices\TechSerGeo\Hydrology\Users\CKubicki\ML_CODE\Correct_Scaler.pkl', 'rb') as f:
    loaded_scaler = pickle.load(f)

X_new_scaled = loaded_scaler.transform(X_new)

with open(r'K:\Mine\TechnicalServices\TechSerGeo\Hydrology\Users\CKubicki\ML_CODE\Correct_data.pkl', 'rb') as f:
    loaded_classifier = pickle.load(f)

y_new_pred = loaded_classifier.predict(X_new_scaled)
```

```python

new_data['PREDICTIONS'] = y_new_pred

new_data['TM_STAMP'] = pd.to_datetime(new_data['TM_STAMP'], unit='s')
```

```python
# Choose where you want the output to be
new_data.to_csv(r'C:\Temp\ML_Reults.csv', index=False)
```
