```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
import matplotlib.pyplot as plt

# Load your dataset
data = pd.read_csv('2yr_raw.txt', delimiter=',')
#data1 = pd.read_excel('Michael_Model_RUN-6-7.xlsx', sheet_name='2Lines-fake_Training_data')
#data = data.append(data1)
data['TM_STAMP'] = (pd.to_datetime(data['TM_STAMP']).astype('int64') // 10**9)
# Now replace NaN values with 0
#data['DIGITS'].fillna(-99, inplace=True)

#data['AMP'].fillna(-9, inplace=True)
data=data.dropna()
```

```python
data
```

```python
# Preprocess the data
from sklearn.utils import shuffle
X = data[['ID_PIEZO','TM_STAMP','DIGITS','AMP']]              #
y = data['REPORTABLE']
```

```python
from sklearn.preprocessing import MinMaxScaler
from imblearn.over_sampling import SMOTE
sm = SMOTE(sampling_strategy='minority')
X , y=sm.fit_resample(X,y)
y.value_counts()
```

```python
scalar=MinMaxScaler()
X1=scalar.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X1, y, test_size=0.3,stratify=y)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(min_samples_leaf=4,max_depth=10,min_samples_split=10,n_estimators=100)
```

```python
rf.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import sklearn.metrics as metrics
X1_pred=rf.predict(X_train)
score2=accuracy_score(y_train,X1_pred)
y1_pred=rf.predict(X_test)
score3=accuracy_score(y_test,y1_pred)
print(score2 , score3)
```

```python
cm = metrics.confusion_matrix(y_test, y1_pred)
```

```python
print(classification_report(y_test, y1_pred))
```

```python
import seaborn as sns
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
```
